Q9: Combining Measures
A family records their monthly electricity usage (in kWh): [400, 420, 390, 450, 410].
Find the mean and standard deviation. What do these values together tell you about the family’s energy use pattern?

  -Ans: First, calculate the mean:
  Mean ={400 + 420 + 390 + 450 + 410}/5 = 414 kWh

Standard Deviation:

Step 1: Find deviations from the mean (414):

* 400-414 → -14
* 420-414 → +6
* 390-414 → -24
* 450-414 → +36
* 410-414 → -4

Step 2: Square deviatins :-14, 6, -24, 36, -4
196, 36, 576, 1296, 16

Step 3: Average of squared deviations:
(196 + 36 + 576 + 1296 + 16)/5 = 424 kwh


Step 4: Standard deviation:
sqrt{424} = approx 20.6 kWh


---

### **Final Answer**

* **Mean usage = 414 kWh**
* **Standard deviation ≈ 20.6 kWh**

---

### **Interpretation**

Together, these values tell you:

* The family typically uses around **414 kWh per month**.
* The **variation is relatively small (≈ 20.6 kWh)** compared to the mean.

[15, 17, 18, 19, 20, 21, 22, 25]

{19 + 20}/2 = 19.5

No number repeats → **No mode**

4. Range
Range = 25 - 15 = 10

5. Standard Deviation
Find Deviation from the mean (19.625)
15 - 19.625 = -4.625
17 - 19.625 = -2.625
18 - 19.625 = -1.625
19 - 19.625 = 0.625
20 - 19.625 = 0.375
21 - 19.625 = 1.375
22 - 19.625 = 2.375
25 - 19.625 = 5.375

Square of deviation and averaje give variance = 8.98

Standard Deviation = sqrt{8.98} = approx 3.0
